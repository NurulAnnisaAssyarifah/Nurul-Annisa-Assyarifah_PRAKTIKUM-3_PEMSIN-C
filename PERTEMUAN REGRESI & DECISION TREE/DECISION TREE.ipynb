## PRE PROCESSING

```python
df = pd.read_csv('iris-flower.csv', sep=",")
df

#202431022_Nurul Annisa Assyarifah
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Species']= le.fit_transform(df['Species'])
print(df['Species'].unique())

for i, label in enumerate(le.classes_):
    print (f"{i} = {label}")
    
#202431022_Nurul Annisa Assyarifah
```

```python
x = df.drop(['Id', 'Species'], axis=1)
x

#202431022_Nurul Annisa Assyarifah
```

```python
y = df['Species']
y

#202431022_Nurul Annisa Assyarifah
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size= 0.25, random_state=42)

#Munculinnya harus urutan karena kalo ga urutan nanti ga bakal kebaca
```

```python
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)

#202431022_Nurul Annisa Assyarifah
```

```python
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize = 10)
plt.show

#Ada beberapa variabel kayak X terus ada value

#202431022_Nurul Annisa Assyarifah
```

```python
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier(criterion='entropy', max_depth=5)
classifier.fit(x_train, y_train)

#202431022_Nurul Annisa Assyarifah
```

```python
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize = 10)
plt.show

#202431022_Nurul Annisa Assyarifah
```

```python
depth = classifier.get_depth()
depth

#202431022_Nurul Annisa Assyarifah
```

```python
#Evaluasi

from sklearn.metrics import accuracy_score

y_pred = classifier.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)
print(akurasi)

#202431022_Nurul Annisa Assyarifah
```
